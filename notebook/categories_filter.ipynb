```python
rootpath = "../Data/"
```

```python
import pandas as pd
```

```python
item_features = pd.read_csv(rootpath + 'item_properties_part1.csv')
```

```python
item_features
```

```python
item_category1 = item_features.loc[item_features.property == 'categoryid', ['itemid', 'value']]
```

```python
item_category1
```

```python
item_features = pd.read_csv(rootpath + 'item_properties_part2.csv')
```

```python
item_category2 = item_features.loc[item_features.property == 'categoryid', ['itemid', 'value']]
```

```python
item_category = pd.concat([item_category1, item_category2], axis=0)
```

```python
item_category.groupby('value').count().sort_values('itemid', ascending=False)
```

```python
item_category[item_category.value == '10']
```

```python
events = pd.read_csv(rootpath + 'events.csv')
events
```

```python
events["event"].unique()
```

```python
events["event"].value_counts()
```

```python
events[events.transactionid.notnull()].visitorid.unique().size
```

```python
events[events.transactionid.notnull()].itemid.unique().size
```

```python
user_item = events[events.transactionid.notnull()][['visitorid', 'itemid']]
```

```python
user_item
```

```python
user_item_category = pd.merge(left = user_item , right = item_category, how = "inner", on = "itemid")
user_item_category['purchase'] = 1
```

```python
user_item_category
```

```python
user_item_category.value.unique().size
```

```python
user_item_category.visitorid.unique().size
```

```python
user_category_matrix = user_item_category.pivot_table(values='purchase', index='visitorid', columns='value', fill_value=0)
```

```python
user_category_matrix.sum().sum()
```

```python
user_category_matrix.shape
```

```python
user_category_matrix
```

```python
from sklearn.decomposition import NMF
```

```python
nmf_model = NMF(n_components=32)

nmf_model.fit(user_category_matrix)
```

```python
user_embedding = nmf_model.transform(user_category_matrix)
category_embedding = nmf_model.components_.T
```

```python
user_embedding[0]
```

```python
category_embedding.shape
```

```python
import numpy as np
```

```python
np.save(rootpath + 'category_embedding.npy', category_embedding)
```

```python
similarities = np.matmul(category_embedding, category_embedding.T)
```

```python
user_category_matrix.columns
```

```python
simlarity_matrix = pd.DataFrame(similarities, index=user_category_matrix.columns, columns=user_category_matrix.columns)
```

```python
simlarity_matrix > 0.01
```

```python
boolean_similarity = simlarity_matrix > 0.01
```
